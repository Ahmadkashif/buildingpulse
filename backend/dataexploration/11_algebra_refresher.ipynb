## Chapter 1: An equation is a balance scale

Forget letters. `=` means **the left side weighs the same as the right side.**

```
    [3 + 2]  =  [5]
```

If you do something to one side, you **must** do the same to the other, or the scale tips.

**The only two moves you ever need:**
1. Add/subtract the same thing to both sides.
2. Multiply/divide both sides by the same (non-zero) thing.

That's it. Every algebra trick is just these two moves repeated.

````python
# Demo: the scale stays balanced
left  = 3 + 2
right = 5
print(f'Start:         {left} = {right}   -> balanced? {left == right}')

# Add 10 to both sides
print(f'Add 10 both:   {left + 10} = {right + 10}   -> balanced? {left + 10 == right + 10}')

# Multiply both by 3
print(f'Times 3 both:  {left * 3} = {right * 3}   -> balanced? {left * 3 == right * 3}')

# Break it: add 10 to left only
print(f'Cheat (add 10 to left only): {left + 10} = {right}   -> balanced? {left + 10 == right}')
````

## Chapter 2: One unknown — finding a missing number

Now replace one number with a letter. The letter is just a placeholder for **a number we don't know yet**.

```
    3·x + 2 = 14
```

Our job: isolate `x` on one side so we can read off the answer.

Think of it like peeling layers off `x`:
- `x` has a `+2` hanging on it → undo with `−2`
- `x` has a `×3` hanging on it → undo with `÷3`

**Always do the undo to BOTH sides** (balance rule).

````python
# Walk through: solve 3x + 2 = 14
print('Step 0:  3·x + 2 = 14')
print()
print('Step 1:  subtract 2 from both sides')
print('         3·x + 2 - 2 = 14 - 2')
print('         3·x         = 12')
print()
print('Step 2:  divide both sides by 3')
print('         3·x / 3 = 12 / 3')
print('         x       = 4')
print()

# Verify
x = 4
print(f'Check:   3·{x} + 2 = {3*x + 2}  (should be 14)  ✓')
````

````python
# Try a few yourself (just run to see solutions)

# Problem A:  5x - 3 = 22
# +3 both sides:  5x = 25
# /5 both sides:  x  = 5
print('A) 5x - 3 = 22  ->  x = 5   check:', 5*5 - 3)

# Problem B:  2x + 7 = 1
# -7 both sides:  2x = -6
# /2 both sides:  x  = -3
print('B) 2x + 7 = 1   ->  x = -3  check:', 2*(-3) + 7)

# Problem C:  x/4 + 1 = 3
# -1 both sides:  x/4 = 2
# *4 both sides:  x   = 8
print('C) x/4 + 1 = 3  ->  x = 8   check:', 8/4 + 1)
````

## Chapter 3: Why two unknowns need two equations

Suppose I tell you:

```
    x + y = 10
```

Can you find `x`? **No.**

- x=1, y=9 works
- x=5, y=5 works
- x=100, y=-90 works

Infinite answers. One equation isn't enough — it only pins down the *relationship* between x and y, not their actual values.

Now give me a second clue:

```
    x + y = 10
    x - y = 4
```

Now `x` and `y` are nailed down. Only **one** pair of numbers satisfies both. (It's x=7, y=3.)

**The rule:** 2 unknowns need 2 equations. 3 unknowns need 3. 83 unknowns need 83.

## Chapter 4: Picture — two lines crossing at one point

Each equation with 2 unknowns is a **line** on a graph (x-axis and y-axis).

- `x + y = 10` is a line: every (x, y) pair on this line satisfies it.
- `x - y = 4`  is another line.

Where do they **cross**? At the one point that satisfies both. That's the answer to the system.

Let's see it.

````python
import numpy as np
import matplotlib.pyplot as plt

x_vals = np.linspace(-2, 12, 100)

# Line 1: x + y = 10  ->  y = 10 - x
y1 = 10 - x_vals

# Line 2: x - y = 4   ->  y = x - 4
y2 = x_vals - 4

fig, ax = plt.subplots(figsize=(8, 8))
ax.plot(x_vals, y1, label='x + y = 10', linewidth=2)
ax.plot(x_vals, y2, label='x - y = 4',  linewidth=2)

# The crossing point
ax.scatter([7], [3], color='red', s=200, zorder=5, label='crossing: (7, 3)')
ax.annotate('(7, 3)', xy=(7, 3), xytext=(7.5, 3.5), fontsize=14, color='red')

ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)
ax.grid(True, alpha=0.3)
ax.legend(fontsize=12)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Each equation = a line. Solution = where they cross.')
ax.set_aspect('equal')
plt.show()

print('The red dot is the ONE (x, y) pair that satisfies both equations.')
print('Our job now: find that point using algebra instead of eyeballing a graph.')
````

## Chapter 5: Method 1 — Elimination (kill one variable)

**The idea in one sentence:** combine the two equations so one variable disappears, leaving you with chapter 2 (one unknown).

**How to make a variable disappear:** arrange for it to have the **same number in front of it** in both equations, then subtract one equation from the other.

Let's do the warm-up system:

```
    x + y = 10    (A)
    x - y = 4     (B)
```

Look at `y`: it's `+y` in A and `-y` in B. If we **add** A and B, the y's cancel.

````python
print('(A)     x + y = 10')
print('(B)     x - y =  4')
print('─────────────────── add them')
print('       2·x + 0 = 14')
print('          2·x  = 14')
print()
print('Now chapter 2 — one unknown:')
print('          x    = 7')
print()
print('Plug x=7 back into (A):')
print('          7 + y = 10')
print('              y = 3')
print()
print('Answer: x = 7, y = 3  (matches the red dot!)')
````

### What if the numbers in front don't already match?

Most of the time they don't. Example:

```
    2x + 3y = 12    (A)
    4x +  y = 14    (B)
```

To kill `x`: make both coefficients of x the same. (A) has 2x, (B) has 4x. Multiply (A) by 2:

```
    (A) × 2 →  4x + 6y = 24      (A′)
    (B)     →  4x +  y = 14      (B)
```

Now subtract (B) from (A′) — the `4x` cancels.

````python
print('(A)        2x + 3y = 12')
print('(B)        4x +  y = 14')
print()
print('Multiply (A) by 2 so both have 4x:')
print('(A′)       4x + 6y = 24')
print('(B)        4x +  y = 14')
print('──────────────────────── subtract (B) from (A′)')
print('           0  + 5y = 10')
print('               5y  = 10')
print('                y  = 2')
print()
print('Plug y=2 into original (A):  2x + 3·2 = 12  ->  2x = 6  ->  x = 3')
print()
print('Check with (B): 4·3 + 2 = 14  ✓')
````

### The elimination recipe

1. Pick a variable to kill.
2. Multiply one (or both) equations so that variable has the **same coefficient** in both.
3. Subtract one equation from the other → that variable is gone.
4. You're left with one equation, one unknown → solve like chapter 2.
5. Plug the answer back into any original equation to find the other variable.

That's it. Five mechanical steps. No creativity required.

## Chapter 6: Method 2 — Substitution (replace one variable)

Different flavor, same destination. Use whichever feels easier on a given problem.

**The idea:** solve one equation for one variable (in terms of the other), then substitute that expression into the other equation.

Example:

```
    x + y = 10    (A)
    x - y = 4     (B)
```

From (A): `y = 10 - x` (just rearrange).

Plug that into (B) wherever you see `y`:

````python
print('From (A):  y = 10 - x')
print()
print('Substitute into (B): x - y = 4')
print('                     x - (10 - x) = 4')
print('                     x - 10 + x   = 4')
print('                         2x - 10  = 4')
print('                             2x   = 14')
print('                              x   = 7')
print()
print('Then y = 10 - 7 = 3.  Same answer.')
````

## Chapter 7: The full problem, solved cleanly

The exact system from our BuildingPulse work:

```
    14·x + 16·y = 48    (A)
    16·x + 21·y = 59    (B)
```

(x and y stand in for `slope1` and `slope2`.)

Plan: kill `x` using elimination.

- (A) has 14 in front of x.
- (B) has 16 in front of x.
- Smallest number both 14 and 16 divide into: 112.
- So multiply (A) by 8, and (B) by 7, to make both `112·x`.

````python
print('Start:')
print('  (A)  14x + 16y = 48')
print('  (B)  16x + 21y = 59')
print()
print('Multiply (A) by 8  → (A′)  112x + 128y = 384')
print('Multiply (B) by 7  → (B′)  112x + 147y = 413')
print()
print('Subtract (A′) from (B′):')
print('  (B′)   112x + 147y = 413')
print('− (A′)   112x + 128y = 384')
print('  ────────────────────────')
print('            0 +  19y =  29')
print()
print('One unknown now (chapter 2 territory):')
print('  19y = 29')
print('    y = 29/19  ≈ 1.526')
print()
print('Plug back into (A):  14x + 16·(29/19) = 48')
print('                     14x + 464/19    = 48')
print('                     14x             = 48 - 464/19')
print('                     14x             = 912/19 - 464/19')
print('                     14x             = 448/19')
print('                       x             = 448 / (14·19)')
print('                       x             = 32/19  ≈ 1.684')
print()
print('Verify with (B):  16·(32/19) + 21·(29/19)')
print('                = 512/19 + 609/19')
print('                = 1121/19')
print('                = 59  ✓')
print()

# Let numpy confirm
import numpy as np
A = np.array([[14, 16], [16, 21]])
b = np.array([48, 59])
sol = np.linalg.solve(A, b)
print(f'Numpy agrees:  x = {sol[0]:.4f},  y = {sol[1]:.4f}')
print(f'Our answer:    x = {32/19:.4f},  y = {29/19:.4f}')
````

## Summary — the whole toolbox

| Situation | Tool |
|---|---|
| 1 unknown, 1 equation | Peel off layers (balance rule) |
| 2 unknowns, 2 equations | **Elimination** or **substitution** |
| 83 unknowns, 83 equations | Same idea, but let a computer do the grinding (that's what matrices are for) |

**The two ideas that carry you through everything:**
1. `=` means balanced — do the same thing to both sides.
2. To solve N unknowns, you need N independent equations, and you collapse them down one variable at a time.

Once this is muscle memory, the matrix version in notebook 10 is literally just a shortcut for what you did here by hand.
