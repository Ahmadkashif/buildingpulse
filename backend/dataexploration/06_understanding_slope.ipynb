# Understanding Slope — From Zero

This notebook teaches you ONE thing: what is a slope and how do we find the best one for a set of points?

## Chapter 1: What is a slope?
Imagine you're walking on a road. The slope tells you how steep it is.

```python
import numpy as np
import matplotlib.pyplot as plt

# A flat road
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot([0, 10], [5, 5], color='saddlebrown', linewidth=4)
ax.plot(1, 5, 'ro', markersize=15)  # you
ax.annotate('You', xy=(1, 5), xytext=(1, 5.8), fontsize=12, ha='center', color='red')
ax.set_xlim(-1, 12)
ax.set_ylim(0, 12)
ax.set_title('A flat road — you walk forward but never go up or down', fontsize=13)
ax.set_xlabel('Distance walked forward')
ax.set_ylabel('Height')
plt.tight_layout()
plt.show()

print('Slope of a flat road = 0')
print('You go forward, but your height never changes.')
```

```python
# An uphill road
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot([0, 10], [2, 7], color='saddlebrown', linewidth=4)
ax.plot(1, 2.5, 'ro', markersize=15)
ax.annotate('You', xy=(1, 2.5), xytext=(1, 3.3), fontsize=12, ha='center', color='red')

# Show the triangle: forward and up
ax.plot([3, 7], [3.5, 3.5], 'b-', linewidth=2)   # forward
ax.plot([7, 7], [3.5, 5.5], 'g-', linewidth=2)    # up
ax.text(5, 2.8, 'forward 4', fontsize=12, color='blue', ha='center')
ax.text(7.5, 4.5, 'up 2', fontsize=12, color='green')

ax.set_xlim(-1, 12)
ax.set_ylim(0, 10)
ax.set_title('An uphill road — you go forward AND up', fontsize=13)
ax.set_xlabel('Distance walked forward')
ax.set_ylabel('Height')
plt.tight_layout()
plt.show()

print('You walked forward 4 steps and went up 2.')
print('Slope = up / forward = 2 / 4 = 0.5')
print()
print('That is ALL a slope is: how much you go UP for each step FORWARD.')
```

```python
# Compare: gentle vs steep
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Gentle
axes[0].plot([0, 10], [2, 4], color='saddlebrown', linewidth=4)
axes[0].plot([2, 8], [2.4, 2.4], 'b-', linewidth=2)
axes[0].plot([8, 8], [2.4, 3.6], 'g-', linewidth=2)
axes[0].text(5, 1.8, 'forward 6', fontsize=11, color='blue', ha='center')
axes[0].text(8.4, 3.0, 'up 1.2', fontsize=11, color='green')
axes[0].set_title(f'Gentle: slope = 1.2/6 = {1.2/6:.2f}', fontsize=13)
axes[0].set_ylim(0, 8)
axes[0].set_xlabel('Forward')
axes[0].set_ylabel('Height')

# Steep
axes[1].plot([0, 10], [1, 7], color='saddlebrown', linewidth=4)
axes[1].plot([2, 8], [2.2, 2.2], 'b-', linewidth=2)
axes[1].plot([8, 8], [2.2, 5.8], 'g-', linewidth=2)
axes[1].text(5, 1.5, 'forward 6', fontsize=11, color='blue', ha='center')
axes[1].text(8.4, 4.0, 'up 3.6', fontsize=11, color='green')
axes[1].set_title(f'Steep: slope = 3.6/6 = {3.6/6:.2f}', fontsize=13)
axes[1].set_ylim(0, 8)
axes[1].set_xlabel('Forward')
axes[1].set_ylabel('Height')

plt.tight_layout()
plt.show()

print('Same forward distance (6), but different heights.')
print('Bigger slope = steeper road.')
```

## Chapter 2: Slope between two points
Forget roads. Now we have data — two buildings.

```python
# Two buildings
sqft_a, eui_a = 10, 50
sqft_b, eui_b = 60, 85

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter([sqft_a, sqft_b], [eui_a, eui_b], color='steelblue', s=120, zorder=5)
ax.plot([sqft_a, sqft_b], [eui_a, eui_b], color='red', linewidth=2)

# Show rise and run
ax.plot([sqft_a, sqft_b], [eui_a, eui_a], 'b--', linewidth=2)  # forward
ax.plot([sqft_b, sqft_b], [eui_a, eui_b], 'g--', linewidth=2)  # up

ax.text(35, 46, f'forward: {sqft_b} - {sqft_a} = {sqft_b - sqft_a}', fontsize=12, color='blue', ha='center')
ax.text(63, 67, f'up: {eui_b} - {eui_a} = {eui_b - eui_a}', fontsize=12, color='green')

ax.annotate(f'Building A\nsqft={sqft_a}, EUI={eui_a}', xy=(sqft_a, eui_a), xytext=(sqft_a+3, eui_a-6), fontsize=10)
ax.annotate(f'Building B\nsqft={sqft_b}, EUI={eui_b}', xy=(sqft_b, eui_b), xytext=(sqft_b-20, eui_b+3), fontsize=10)

ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('Slope between two buildings', fontsize=13)
plt.tight_layout()
plt.show()

run = sqft_b - sqft_a
rise = eui_b - eui_a
slope = rise / run

print(f'Forward (run):  {sqft_b} - {sqft_a} = {run}')
print(f'Up (rise):      {eui_b} - {eui_a} = {rise}')
print(f'Slope:          {rise} / {run} = {slope:.2f}')
print()
print(f'For every 1 sqft bigger, EUI goes up by {slope:.2f}')
```

## Chapter 3: The problem — more than 2 points
With 2 points, there's exactly one line. With more points, different pairs give different slopes.

```python
# 6 buildings
sqft = np.array([10, 20, 30, 40, 50, 60], dtype=float)
eui  = np.array([50, 55, 65, 70, 80, 85], dtype=float)

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)
for i in range(len(sqft)):
    ax.annotate(f'  B{i+1}', xy=(sqft[i], eui[i]), fontsize=10)

ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('6 buildings — which line fits best?', fontsize=13)
plt.tight_layout()
plt.show()

print('6 buildings. They roughly go up-and-to-the-right, but not perfectly.')
```

```python
# Let's check the slope between each neighboring pair
fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)

colors = ['#e41a1c', '#ff7f00', '#4daf4a', '#377eb8', '#984ea3']

print('Slope between each pair of neighbors:')
print()
for i in range(len(sqft) - 1):
    rise = eui[i+1] - eui[i]
    run = sqft[i+1] - sqft[i]
    slope = rise / run
    ax.plot([sqft[i], sqft[i+1]], [eui[i], eui[i+1]], color=colors[i], linewidth=2, 
            label=f'B{i+1}->B{i+2}: slope={slope}')
    print(f'  B{i+1} -> B{i+2}:  up {rise:.0f} / forward {run:.0f} = slope {slope:.1f}')

ax.legend(fontsize=10)
ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('Each pair gives a DIFFERENT slope', fontsize=13)
plt.tight_layout()
plt.show()

print()
print('We get 0.5 and 1.0 — different slopes depending on which pair we pick.')
print('We need ONE number that represents the best slope for ALL 6 points.')
```

## Chapter 4: The center of the data
Instead of comparing pairs, we compare every point to the CENTER — the average sqft and average EUI.

```python
# Find the center
avg_sqft = sqft.mean()
avg_eui = eui.mean()

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

for i in range(len(sqft)):
    ax.annotate(f'  B{i+1}', xy=(sqft[i], eui[i]), fontsize=10)

ax.annotate(f'  CENTER\n  ({avg_sqft}, {avg_eui})', xy=(avg_sqft, avg_eui),
            fontsize=11, color='red', fontweight='bold')

ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('The red X is the center — average of all sqft, average of all EUI', fontsize=13)
plt.tight_layout()
plt.show()

print(f'Average sqft: ({" + ".join(str(int(s)) for s in sqft)}) / 6 = {avg_sqft}')
print(f'Average EUI:  ({" + ".join(str(int(e)) for e in eui)}) / 6 = {avg_eui}')
print()
print(f'Center = ({avg_sqft}, {avg_eui})')
```

## Chapter 6: Combining all rises and runs into ONE slope
We have 6 rises and 6 runs. We need one single slope number.

```python
# Calculate run and rise for every building
runs  = sqft - avg_sqft
rises = eui - avg_eui

print('Each building\'s run and rise from the center:')
print()
print('Building |  sqft  |  EUI   |  run   |  rise')
print('-' * 52)
for i in range(len(sqft)):
    print(f'   B{i+1}    | {sqft[i]:5.0f}  | {eui[i]:5.0f}  | {runs[i]:+6.1f} | {rises[i]:+6.1f}')
```

```python
# PROBLEM: Why can't we just do total rise / total run?

total_rise = rises.sum()
total_run  = runs.sum()

print('What if we just add up all the rises and all the runs?')
print()
print(f'Total rise = {" + ".join(f"{r:+.1f}" for r in rises)} = {total_rise:.1f}')
print(f'Total run  = {" + ".join(f"{r:+.1f}" for r in runs)} = {total_run:.1f}')
print()
print(f'Both are ZERO! The negatives cancel out the positives.')
print(f'We can\'t divide by zero. This approach fails.')
print()
print('We need a way to combine them WITHOUT the cancellation.')
```

```python
# SOLUTION Part 1: Multiply each building's run x rise
#
# Why multiply?
# If run is negative AND rise is negative:  negative x negative = POSITIVE
# If run is positive AND rise is positive:  positive x positive = POSITIVE
# They agree! And multiplying preserves that.

products = runs * rises

print('Multiply each building\'s run x rise:')
print()
print('Building |  run   |  rise  |  run x rise  | why positive?')
print('-' * 72)
for i in range(len(sqft)):
    reason = 'both negative' if runs[i] < 0 else 'both positive'
    print(f'   B{i+1}    | {runs[i]:+6.1f} | {rises[i]:+6.1f} | {products[i]:+10.2f}   | {reason}')

print(f'\nSum of all products = {products.sum():.1f}')
print()
print('All products are positive — every building agrees the trend is upward.')
print('This sum is the TOTAL RISE CONTRIBUTION. It goes on TOP of our fraction.')
```

```python
# SOLUTION Part 2: Square each run
#
# We need to divide by something that measures "how much total forward movement."
# But runs have negatives that cancel. Squaring makes them all positive.
#   (-25)^2 = 625    (positive!)
#   (+25)^2 = 625    (positive!)

runs_squared = runs ** 2

print('Square each building\'s run:')
print()
print('Building |  run   |  run squared')
print('-' * 38)
for i in range(len(sqft)):
    print(f'   B{i+1}    | {runs[i]:+6.1f} | {runs_squared[i]:8.1f}')

print(f'\nSum of squared runs = {runs_squared.sum():.1f}')
print()
print('No cancellation! This measures the total horizontal spread.')
print('This goes on the BOTTOM of our fraction.')
```

```python
# THE ANSWER: divide top by bottom

top = products.sum()
bottom = runs_squared.sum()
best_slope = top / bottom

print('=' * 50)
print('THE BEST SLOPE')
print('=' * 50)
print()
print(f'        total rise contribution     {top:.1f}')
print(f'slope = ----------------------- = -------- = {best_slope:.4f}')
print(f'        total horizontal spread     {bottom:.1f}')
print()
print(f'For every 1 sqft increase, EUI goes up by {best_slope:.4f}')
print()
print('This is still just rise / run.')
print('The only trick was using multiplication and squaring')
print('to avoid the cancellation problem.')
```

## Chapter 7: Does this slope actually fit the data?
Let's draw the line and see.

```python
# The intercept: where the line crosses the y-axis
# Our line must pass through the center (avg_sqft, avg_eui).
# line formula: EUI = slope * sqft + intercept
# Plug in the center: avg_eui = slope * avg_sqft + intercept
# So: intercept = avg_eui - slope * avg_sqft

intercept = avg_eui - best_slope * avg_sqft

print(f'The line passes through the center ({avg_sqft}, {avg_eui})')
print(f'intercept = {avg_eui} - {best_slope:.4f} x {avg_sqft} = {intercept:.2f}')
print()
print(f'Our line: EUI = {best_slope:.4f} x sqft + {intercept:.2f}')
```

```python
# Draw it!
fig, ax = plt.subplots(figsize=(8, 5))

# The data points
ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)

# Our best fit line
x_line = np.linspace(0, 70, 100)
y_line = best_slope * x_line + intercept
ax.plot(x_line, y_line, color='red', linewidth=2, label=f'slope={best_slope:.3f}')

# The center
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

# Show errors as gray dashed lines
for i in range(len(sqft)):
    predicted = best_slope * sqft[i] + intercept
    ax.plot([sqft[i], sqft[i]], [eui[i], predicted], 'gray', linestyle='--', linewidth=1)
    ax.annotate(f'  B{i+1}', xy=(sqft[i], eui[i]), fontsize=9)

ax.legend(fontsize=12)
ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('Our best fit line — gray dashes show the errors', fontsize=13)
plt.tight_layout()
plt.show()

print('The line passes through the center and fits the points closely.')
print('Gray dashes = how far off each prediction is. Shorter = better.')
```

```python
# Prove it: compare our slope to some random guesses
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
x_line = np.linspace(0, 70, 100)

test_slopes = [
    (0.3, 'Too gentle'),
    (1.2, 'Too steep'),
    (0.5, 'Close but not best'),
    (best_slope, 'OUR BEST SLOPE'),
]

for ax, (s, label) in zip(axes.flat, test_slopes):
    # Intercept so line passes through center
    b = avg_eui - s * avg_sqft
    predictions = s * sqft + b
    total_error = ((eui - predictions) ** 2).sum()
    
    ax.scatter(sqft, eui, color='steelblue', s=80, zorder=5)
    ax.plot(x_line, s * x_line + b, color='red', linewidth=2)
    ax.scatter(avg_sqft, avg_eui, color='red', s=150, marker='X', zorder=5)
    
    for i in range(len(sqft)):
        ax.plot([sqft[i], sqft[i]], [eui[i], predictions[i]], 'gray', linestyle='--', linewidth=1)
    
    color = 'green' if s == best_slope else 'black'
    ax.set_title(f'{label}\nslope={s:.4f}, total error={total_error:.1f}', fontsize=11, color=color)
    ax.set_xlabel('sqft')
    ax.set_ylabel('EUI')
    ax.set_ylim(30, 100)

plt.tight_layout()
plt.show()

print('Look at the total error numbers in each title.')
print('Our slope has the LOWEST error. No other slope can beat it.')
```

## Summary

1. **Slope** = rise / run = how much Y changes per unit of X
2. With 2 points, there's one slope. With 6 points, different pairs give different slopes.
3. To find the BEST slope for all points:
   - Pick a center point (the averages)
   - Measure each point's rise and run from the center
   - **Top of fraction:** multiply each point's rise x run, add them up
   - **Bottom of fraction:** square each point's run, add them up
   - **Divide** = best slope
4. This is still rise / run — just computed across all points at once, using multiplication and squaring to avoid the cancellation problem.
