# How Each Building Votes For a Slope

Each building has its own opinion about what the slope should be. We combine their votes into one answer.

```python
import numpy as np
import matplotlib.pyplot as plt

sqft = np.array([10, 20, 30, 40, 50, 60], dtype=float)
eui  = np.array([50, 55, 65, 70, 80, 85], dtype=float)

avg_sqft = sqft.mean()  # 35
avg_eui = eui.mean()    # 67.5
```

## Building 1 votes

```python
i = 0
run_i = sqft[i] - avg_sqft
rise_i = eui[i] - avg_eui
vote_i = rise_i / run_i

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(sqft, eui, color='lightgray', s=80, zorder=3)
ax.scatter(sqft[i], eui[i], color='steelblue', s=120, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

# Line from center through this point
x_line = np.linspace(0, 70, 100)
ax.plot(x_line, vote_i * (x_line - avg_sqft) + avg_eui, color='steelblue', linewidth=2, alpha=0.7)

# Rise and run arrows
ax.annotate('', xy=(sqft[i], avg_eui), xytext=(avg_sqft, avg_eui),
            arrowprops=dict(arrowstyle='->', color='blue', lw=2))
ax.annotate('', xy=(sqft[i], eui[i]), xytext=(sqft[i], avg_eui),
            arrowprops=dict(arrowstyle='->', color='green', lw=2))

ax.text(22, 69, f'run = {run_i:.0f}', fontsize=12, color='blue')
ax.text(11, 57, f'rise = {rise_i:.1f}', fontsize=12, color='green')

ax.set_title(f'Building 1 votes: slope = {rise_i:.1f} / {run_i:.0f} = {vote_i:.3f}', fontsize=13)
ax.set_xlabel('sqft')
ax.set_ylabel('EUI')
ax.set_xlim(-5, 75)
ax.set_ylim(35, 95)
plt.tight_layout()
plt.show()

print(f'B1 is at sqft={sqft[i]:.0f}, EUI={eui[i]:.0f}')
print(f'Center is at sqft={avg_sqft}, EUI={avg_eui}')
print(f'Run = {run_i:.0f} (25 to the LEFT)')
print(f'Rise = {rise_i:.1f} (17.5 BELOW)')
print(f'Slope = {rise_i:.1f} / {run_i:.0f} = {vote_i:.3f}')
print(f'\nThe blue line shows what the slope would look like if B1 got to decide alone.')
```

## Building 2 votes

```python
i = 1
run_i = sqft[i] - avg_sqft
rise_i = eui[i] - avg_eui
vote_i = rise_i / run_i

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(sqft, eui, color='lightgray', s=80, zorder=3)
ax.scatter(sqft[i], eui[i], color='steelblue', s=120, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

x_line = np.linspace(0, 70, 100)
ax.plot(x_line, vote_i * (x_line - avg_sqft) + avg_eui, color='steelblue', linewidth=2, alpha=0.7)

ax.annotate('', xy=(sqft[i], avg_eui), xytext=(avg_sqft, avg_eui),
            arrowprops=dict(arrowstyle='->', color='blue', lw=2))
ax.annotate('', xy=(sqft[i], eui[i]), xytext=(sqft[i], avg_eui),
            arrowprops=dict(arrowstyle='->', color='green', lw=2))

ax.text(27, 69, f'run = {run_i:.0f}', fontsize=12, color='blue')
ax.text(21, 60, f'rise = {rise_i:.1f}', fontsize=12, color='green')

ax.set_title(f'Building 2 votes: slope = {rise_i:.1f} / {run_i:.0f} = {vote_i:.3f}', fontsize=13)
ax.set_xlabel('sqft')
ax.set_ylabel('EUI')
ax.set_xlim(-5, 75)
ax.set_ylim(35, 95)
plt.tight_layout()
plt.show()

print(f'B2 wants a steeper slope ({vote_i:.3f}) than B1 (0.700).')
print(f'Its line through the center would be steeper.')
```

## Building 3 votes

```python
i = 2
run_i = sqft[i] - avg_sqft
rise_i = eui[i] - avg_eui
vote_i = rise_i / run_i

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(sqft, eui, color='lightgray', s=80, zorder=3)
ax.scatter(sqft[i], eui[i], color='steelblue', s=120, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

x_line = np.linspace(0, 70, 100)
ax.plot(x_line, vote_i * (x_line - avg_sqft) + avg_eui, color='steelblue', linewidth=2, alpha=0.7)

ax.annotate('', xy=(sqft[i], avg_eui), xytext=(avg_sqft, avg_eui),
            arrowprops=dict(arrowstyle='->', color='blue', lw=2))
ax.annotate('', xy=(sqft[i], eui[i]), xytext=(sqft[i], avg_eui),
            arrowprops=dict(arrowstyle='->', color='green', lw=2))

ax.text(32, 69, f'run = {run_i:.0f}', fontsize=12, color='blue')
ax.text(31, 64, f'rise = {rise_i:.1f}', fontsize=12, color='green')

ax.set_title(f'Building 3 votes: slope = {rise_i:.1f} / {run_i:.0f} = {vote_i:.3f}', fontsize=13)
ax.set_xlabel('sqft')
ax.set_ylabel('EUI')
ax.set_xlim(-5, 75)
ax.set_ylim(35, 95)
plt.tight_layout()
plt.show()

print(f'B3 wants a gentler slope ({vote_i:.3f}).')
print(f'But notice — B3 is very close to the center.')
print(f'Its run is only {abs(run_i):.0f}. Small run = unreliable vote.')
```

## All 6 buildings vote — see their lines

```python
runs  = sqft - avg_sqft
rises = eui - avg_eui
votes = rises / runs

colors = ['#e41a1c', '#ff7f00', '#4daf4a', '#377eb8', '#984ea3', '#a65628']

fig, ax = plt.subplots(figsize=(10, 6))
x_line = np.linspace(0, 70, 100)

for i in range(len(sqft)):
    ax.plot(x_line, votes[i] * (x_line - avg_sqft) + avg_eui, 
            color=colors[i], linewidth=1.5, alpha=0.6,
            label=f'B{i+1}: slope={votes[i]:.3f}')

ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

for i in range(len(sqft)):
    ax.annotate(f' B{i+1}', xy=(sqft[i], eui[i]), fontsize=10)

ax.legend(fontsize=10, loc='upper left')
ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('Each building wants a different line — 6 different slopes', fontsize=13)
ax.set_xlim(-5, 75)
ax.set_ylim(35, 95)
plt.tight_layout()
plt.show()

print('Each colored line is what ONE building would choose.')
print('They all pass through the center but have different steepness.')
print('We need to pick ONE slope that best represents all of them.')
```

## Why not just average the votes equally?

```python
simple_avg = votes.mean()

print('All votes:')
print()
for i in range(len(sqft)):
    dist = abs(runs[i])
    print(f'  B{i+1}: slope = {votes[i]:.3f}   (distance from center: {dist:.0f})')

print(f'\nSimple average: {simple_avg:.3f}')
print()
print('But look at the distances!')
print('B3 is only 5 away from center. B1 is 25 away.')
print()
print('B3 is like someone squinting at something 2 feet away — tiny')
print('movements look huge. Its slope measurement is shaky.')
print()
print('B1 is like someone looking from far away — it has a wide view')
print('and a more reliable sense of the overall trend.')
print()
print('Should they get the same vote? No!')
```

## Weighted voting: further buildings count more

```python
# Weight = how far the building is from center (squared)
# Further away = bigger weight = louder voice

weights = runs ** 2  # distance squared

fig, ax = plt.subplots(figsize=(8, 4))
bars = ax.bar([f'B{i+1}' for i in range(len(sqft))], weights, color=colors)
ax.set_ylabel('Vote weight (distance² from center)', fontsize=11)
ax.set_title('How loud each building\'s vote is', fontsize=13)

for i, (bar, w) in enumerate(zip(bars, weights)):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 10,
            f'{w:.0f}', ha='center', fontsize=11)

plt.tight_layout()
plt.show()

print('B1 and B6 are furthest from center — they get the loudest voice (625).')
print('B3 and B4 are closest — they barely get a say (25).')
```

```python
# Weighted average: each vote x its weight, then divide by total weight

print('Weighted voting:')
print()
print('Building | vote  | weight | vote x weight')
print('-' * 48)

weighted_products = votes * weights
for i in range(len(sqft)):
    print(f'  B{i+1}     | {votes[i]:.3f} | {weights[i]:6.0f} | {weighted_products[i]:8.2f}')

print(f'\nSum of (vote x weight): {weighted_products.sum():.2f}')
print(f'Sum of weights:         {weights.sum():.0f}')
print(f'\nWeighted slope = {weighted_products.sum():.2f} / {weights.sum():.0f} = {weighted_products.sum() / weights.sum():.4f}')
```

```python
# This IS the formula!
# vote x weight = (rise/run) x run² = rise x run
# sum of weights = sum of run²
#
# So: weighted average = sum(rise x run) / sum(run²)
# That's EXACTLY the formula we've been using!

formula_slope = (rises * runs).sum() / (runs ** 2).sum()

print('Here\'s the secret:')
print()
print('  vote x weight = (rise / run) x run²')
print('                = rise x run')
print()
print('  sum of weights = sum of run²')
print()
print('  So the weighted average = sum(rise x run) / sum(run²)')
print()
print('  That is EXACTLY the slope formula!')
print()
print(f'  From voting:  {weighted_products.sum() / weights.sum():.4f}')
print(f'  From formula: {formula_slope:.4f}')
print()
print('  Same number. The formula is just a weighted vote in disguise.')
```

## The final line

```python
best_slope = formula_slope
intercept = avg_eui - best_slope * avg_sqft

fig, ax = plt.subplots(figsize=(10, 6))
x_line = np.linspace(0, 70, 100)

# Faded: each building's individual vote line
for i in range(len(sqft)):
    ax.plot(x_line, votes[i] * (x_line - avg_sqft) + avg_eui,
            color=colors[i], linewidth=1, alpha=0.25)

# Bold: the weighted average (best slope)
ax.plot(x_line, best_slope * x_line + intercept, color='black', linewidth=3,
        label=f'Best slope = {best_slope:.4f} (weighted average)')

ax.scatter(sqft, eui, color='steelblue', s=100, zorder=5)
ax.scatter(avg_sqft, avg_eui, color='red', s=200, marker='X', zorder=5)

for i in range(len(sqft)):
    ax.annotate(f' B{i+1}', xy=(sqft[i], eui[i]), fontsize=10)

ax.legend(fontsize=12)
ax.set_xlabel('sqft', fontsize=12)
ax.set_ylabel('EUI', fontsize=12)
ax.set_title('Faded = each building\'s vote. Bold = the weighted average.', fontsize=13)
ax.set_xlim(-5, 75)
ax.set_ylim(35, 95)
plt.tight_layout()
plt.show()

print('The black line is the compromise.')
print('Buildings far from center had the loudest voice.')
print('Buildings close to center barely influenced it.')
print(f'\nFinal slope: {best_slope:.4f}')
```

## Summary

1. Each building draws a line from the center through itself — that's its **vote** for the slope.
2. Buildings far from the center get a **louder vote** (they have more evidence).
3. Buildings close to the center get a **quiet vote** (small rise and run = unreliable).
4. The best slope is the **weighted average** of all votes.
5. The formula `sum(rise x run) / sum(run²)` is just a shortcut for this weighted vote.
